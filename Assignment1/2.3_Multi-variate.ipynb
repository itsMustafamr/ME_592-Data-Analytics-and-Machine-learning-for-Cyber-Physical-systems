```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python

col_names = [
    "frequency",
    "attack_angle",
    "chord_length",
    "free_stream_velocity",
    "suction_side_displacement_thickness",
    "scaled_sound_pressure_level"
]

df = pd.read_csv(
    "airfoil_self_noise.dat",  
    sep="\t",
    header=None,
    names=col_names
)
print(df.head())
```

```python

# ── Descriptive Statistics ────────────────────────────────────────────────────
print("=" * 75)
print("AIRFOIL SELF-NOISE DATASET — DESCRIPTIVE STATISTICS")
print("=" * 75)
print(f"\nDataset shape: {df.shape[0]} instances × {df.shape[1]} variables\n")
```

```python

stats_dict = {}
for col in df.columns:
    s = df[col]
    stats_dict[col] = {
        "Mean":      round(s.mean(), 4),
        "Variance":  round(s.var(), 4),
        "Std Dev":   round(s.std(), 4),
        "Median":    round(s.median(), 4),
        "Kurtosis":  round(s.kurtosis(), 4),   # excess kurtosis (normal = 0)
        "Skewness":  round(s.skew(), 4),
        "Range":     round(s.max() - s.min(), 4),
    }

results = pd.DataFrame(stats_dict).T
results.index.name = "Variable"

# Pretty-print
pd.set_option("display.max_columns", 10)
pd.set_option("display.width", 120)
pd.set_option("display.float_format", "{:,.4f}".format)
print(results.to_string())

# ── Per-variable detailed breakdown ──────────────────────────────────────────
print("\n" + "=" * 75)
print("DETAILED BREAKDOWN PER VARIABLE")
print("=" * 75)
for col, s_stats in stats_dict.items():
    print(f"\n► {col.replace('_', ' ').title()}")
    for stat_name, val in s_stats.items():
        print(f"   {stat_name:<12}: {val:>14,.4f}")
```

```python

print("\n[Note] Kurtosis is reported as *excess* kurtosis (Normal distribution = 0).")
print("       Positive kurtosis → heavier tails than normal.")
print("       Negative kurtosis → lighter tails than normal.")
```
