## Classes

```python
class Model(object):
    # class initialization and attributes
    def __init__(self, params, model_type, name='Neural Network'):
        self.name = name
        self.params = params
        self.model_type = model_type

    # internal functions denoted by __.__
    # str
    def __str__(self):
        return 'Readable string : This is a '+ self.model_type + ' ' + self.name + ' with ' + str(self.params) + ' params'

    # representation
    def __repr__(self):
        return self.model_type + ' ' + self.name + ' with ' + str(self.params) + ' params'

    # method
    def printParams(self, multiplier):
        print(self.params*multiplier)
```

```python
m1 = Model(5, 'Supervised', 'LinearRegression')
```

```python
print(m1)
```

```python
import numpy as np
import scipy.signal
import matplotlib.pyplot as plt

# 1. Ensure data is numeric (fixes the ValueError)
signal_to_filter = timedata_np[:, 0].astype(float)

def apply_convolution(sig, window):
    # Create a proper Moving Average kernel: [1, 1, 1, ...]
    # A kernel of all 1s of length 'window'
    conv = np.ones(window)

    # Normalize the kernel so the signal amplitude doesn't change
    conv = conv / window

    print('Convolution Kernel shape:', conv.shape)

    # Apply convolution
    filtered = scipy.signal.convolve(sig, conv, mode='same')
    return filtered

# 2. Apply the filters to the cleaned numeric data
filtered_med = scipy.signal.medfilt(signal_to_filter, 7)
convolved_avg = apply_convolution(signal_to_filter, 20)

# 3. Plotting
plt.plot(signal_to_filter, label='Original', alpha=0.3)
plt.plot(filtered_med, label='Median Filter (k=7)')
plt.legend()
plt.show()


plt.plot(signal_to_filter, label='Original', alpha=0.3)
plt.plot(convolved_avg, label='Moving Average (w=20)', color='orange')
plt.legend()
plt.show()
```

mean is the average. Variance is “how much things spread out around the mean.”

1) Mean vs Variance (simple)

Say you have numbers: 2, 4, 6

Mean (average) = 
(2+4+6)/3=4

Variance measures how far values are from 4, on average.

Distances from mean: 
−2,0,+2
Squares: 

4,0,4
Average of squares = 
(4+0+4)/3=2.67

So:

Small variance → values are clustered near the mean

Large variance → values are spread out

Variance is basically a measure of spread.

```python
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
```

```python
pca = PCA(n_components=2)
pca_result = pca.fit_transform(images.reshape(10000,28*28))
pca_result.shape
```

```python
plt.figure()
plt.scatter(pca_result[:,0], pca_result[:,1], c=labels)
plt.colorbar()
plt.show()
```

this is exactly where PCA sounds like “image compression,” but what it’s doing here is not turning a 28×28 digit into a sharper 2×2 digit.

1) “Preserves maximum variance” — what that means

In PCA, variance means: how spread out the data is along some direction.

Imagine you have a bunch of points in a high-dimensional space (here: each image is a point in 784-D because 28×28 = 784 pixels).

PCA looks for a direction (a line through that space) such that:

if you project all points onto that line,

the projected values are as spread out as possible.

That “spread out as possible” is maximum variance.

Why is that good?

Directions with high variance capture big differences between images.

Directions with low variance often capture small/noisy variations.

So PCA chooses:

PC1: direction with the most variance

PC2: direction with the next most variance, but orthogonal to PC1

etc.

```python
tsne = TSNE(n_components=2, verbose=1, perplexity=40, n_iter=300)
import time
start = time.time()
tsne_results = tsne.fit_transform(images.reshape(10000,28*28))
end = time.time()
print('Time', end-start)
```

```python
plt.scatter(tsne_results[:,0], tsne_results[:,1], c=labels)
plt.colorbar()
plt.show()
```

```python
# applying PCA n high-dimensional data, then using t-SNE can speed up process
pca = PCA(n_components=50)
pca_result_int = pca.fit_transform(images.reshape(10000,28*28))
start = time.time()
tsne = TSNE(n_components=2, verbose=1, perplexity=40, n_iter=300)
tsne_results = tsne.fit_transform(pca_result_int)
end = time.time()
print('Time', end-start)
```
